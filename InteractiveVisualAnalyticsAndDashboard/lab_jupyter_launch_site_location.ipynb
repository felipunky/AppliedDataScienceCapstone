```python
import numpy as np
from sklearn.cluster import KMeans
from math import radians, sin, cos, sqrt, atan2

# Haversine distance on the sphere (great-circle distance in km)
def haversine_km(lat1, lon1, lat2, lon2):
    R = 6371.0
    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = sin(dlat / 2) ** 2 + cos(lat1) * cos(lat2) * sin(dlon / 2) ** 2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    return R * c

# Convert lat/lon into 3D unit-sphere coordinates so Euclidean K-means behaves like
# a spherical clustering method consistent with Haversine distance.
coords = launch_sites_df[['Lat', 'Long']].to_numpy()
X = np.column_stack([
    np.cos(np.radians(coords[:, 0])) * np.cos(np.radians(coords[:, 1])),
    np.cos(np.radians(coords[:, 0])) * np.sin(np.radians(coords[:, 1])),
    np.sin(np.radians(coords[:, 0]))
])

kmeans = KMeans(n_clusters=4, random_state=0, n_init=10)
launch_sites_df['cluster'] = kmeans.fit_predict(X)

# show the cluster assignment for each launch site
launch_sites_df[['Launch Site', 'Lat', 'Long', 'cluster']]
```

```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

*TODO:* Mark down a point on the closest coastline using MousePosition and calculate the distance between the coastline point and the launch site.


```python
# find coordinate of the closest coastline
# Example: use a coastline point near the selected launch site
coastline_point = [28.56367, -80.57163]
coastline_lat, coastline_lon = coastline_point

# Use one launch site to compute a scalar distance
launch_site = launch_sites_df.iloc[0]
distance_coastline = calculate_distance(
    launch_site['Lat'],
    launch_site['Long'],
    coastline_lat,
    coastline_lon
)

print(f"Distance to coastline: {distance_coastline:.2f} km")
```

```python
# Create and add a folium.Marker on your selected closest coastline point on the map
# Display the distance between coastline point and launch site using the icon property
# for example

distance_marker = folium.Marker(
    coastline_point,
    icon=DivIcon(
        icon_size=(20, 20),
        icon_anchor=(0, 0),
        html='<div style="font-size: 12; color:#d35400;"><b>{:.2f} KM</b></div>'.format(distance_coastline),
    )
)
site_map.add_child(distance_marker)
site_map
```
